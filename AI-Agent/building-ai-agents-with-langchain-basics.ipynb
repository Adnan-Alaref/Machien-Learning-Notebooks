# **Step 1: What is an AI Agent?**

An **AI Agent** is an AI system that can **perceive information**, **make decisions**, and **take actions** to achieve a specific __goal__.

Unlike a simple chatbot that only answers questions, an AI agent can:

- 📋 Plan tasks
- 🛠️ Use external tools
- 🧠 Remember information
- ⚡ Perform actions on your behalf

---

# 🏗️ Core Components of an AI Agent

## 🎯 1. Goal

Defines what the agent is trying to achieve.

**Example:**
> "Book the cheapest flight to Cairo."

---

## 👀 2. Perception

The agent receives information from its environment.

**Examples:**

- Reading user messages
- Browsing websites
- Accessing databases
- Processing emails

---

## 🧩 3. Reasoning & Planning

The agent determines the steps needed to accomplish its goal.

**Example Workflow:**

1. Search for available flights
2. Compare prices
3. Select the best option
4. Present or book the flight

---

## ⚙️ 4. Actions

The agent executes tasks using available tools.

**Examples:**

- 🌐 Search the web
- 💻 Run code
- 📧 Send emails
- 📊 Update spreadsheets
- 📅 Schedule meetings

---

## 🧠 5. Memory *(Optional)*

Stores information for future interactions and better personalization.

**Example:**
> Remembers your preferred airline and always checks it first when searching for flights.

---

# 💡 Simple Comparison

| Chatbot | AI Agent |
|----------|----------|
| Answers questions | Achieves goals |
| Responds to prompts | Plans and acts |
| Limited memory | Can remember information |
| No tool usage | Can use tools |
| Reactive | Proactive |

---
# 🤖 Types of AI Agents

AI agents can be classified based on how they make decisions and interact with their environment.

| Type | Description |
|--------|-------------|
| **Reactive Agent** | Responds directly to inputs without planning or memory. |
| **Goal-Based Agent** | Chooses actions that help achieve a specific goal. |
| **Utility-Based Agent** | Selects the action with the highest expected benefit or utility. |
| **Learning Agent** | Improves its performance through experience and feedback. |
| **Multi-Agent System** | Multiple agents collaborate or compete to accomplish tasks. |

---

# 🧩 PromptTemplate in LangChain

## What is a PromptTemplate?

- A **PromptTemplate** is a reusable prompt with placeholders (variables) that are filled with values at runtime.

- Also, **PromptTemplate** is used to create __dynamic prompts__ by inserting variables into a text template before sending it to an __LLM__.

- Instead of **writing** a new __prompt every time__, you create a __template once__ and provide different inputs when needed.

### Without PromptTemplate

```python
topic = "Machine Learning"

prompt = f"""
Explain {topic} in simple terms.
"""
```

This works, but becomes difficult to manage when prompts get larger.

---

### With PromptTemplate

```python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms."
)

prompt = prompt_template.format(topic="Machine Learning")

print(prompt)
```

**Output**

```text
Explain Machine Learning in simple terms.
```

---

## Why Use PromptTemplate?

### 1️⃣ Reusability

Create the prompt once and use it many times.

```python
prompt_template.format(topic="Python")
prompt_template.format(topic="Deep Learning")
prompt_template.format(topic="Data Science")
```

---

### 2️⃣ Cleaner Code

Instead of mixing text and variables with f-strings:

```python
prompt = f"""
You are a teacher.
Explain {topic}.
Give 3 examples.
"""
```

Use:

```python
prompt_template = PromptTemplate(
    input_variables=["topic"],
    template="""
    You are a teacher.
    Explain {topic}.
    Give 3 examples.
    """
)
```

---

### 3️⃣ Easier Maintenance

If you want to change the prompt, edit the template once instead of modifying multiple f-strings across your project.

---

## Components of PromptTemplate

```python
PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms."
)
```

### `input_variables`

A list containing all variables used inside the prompt.

```python
input_variables=["topic"]
```

### `template`

The actual prompt text.

```python
template="Explain {topic} in simple terms."
```

### Placeholder

Variables inside `{}` are placeholders.

```python
{topic}
```
---

### Workflow

```text

     ▼

     ▼
```

# Summary

| Concept | Description |
|----------|------------|
| PromptTemplate | A reusable prompt with variables |
| input_variables | Variables that will be filled later |
| template | Prompt text containing placeholders |
| format() | Replaces placeholders with actual values |
| Main Benefit | Cleaner, reusable, maintainable prompts |

### Mental Model

```text
Template + Variables
        │
        ▼
   PromptTemplate
        │
        ▼
   Final Prompt
        │
        ▼
       LLM
        │
        ▼
     Response
```

Think of a **PromptTemplate** as a form:

```text
Hello {name},
Welcome to {company}.
```

Fill the blanks:

```text
name = Adnan
company = OpenAI
```

Result:

```text
Hello Adnan,
Welcome to OpenAI.
```

This is exactly what PromptTemplate does before sending the prompt to the LLM.

````python
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI

prompt = PromptTemplate(
  input_variables=["student_level", "topic"],
  template="""
  You are an expert tutor
  Student level: {student_level}
  Teach the following topic: {topic}
  Use examples and keep explanations appropriate for the student's level.
  """
)

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash",
                             google_api_key = google_api_key,
                             temperature=0)

chain = prompt | llm

response = chain.invoke({
    "student_level": "Beginner",
    "topic": "Neural Networks"
})

display(Markdown(response.content))
````
